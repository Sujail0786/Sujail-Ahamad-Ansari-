# Medical Data Analysis: Patient Health and Diabetes Risk Patterns

**Project type:** Educational exploratory data analysis (EDA)  
**Dataset:** `Medical_Dataset.csv` (synthetic data; 500 records)

## Project prompt
You are a junior data analyst working with a fictional health research team. The team wants to understand patterns in a synthetic patient dataset and explore how age, BMI, glucose, blood pressure, cholesterol, physical activity and family history vary across two simulated outcome groups.

**Tasks**
1. Load and inspect the dataset.
2. Check data types, missing values and duplicate records.
3. Clean missing numeric values using median imputation.
4. Calculate descriptive statistics and compare group averages.
5. Create visualizations: outcome distribution, age distribution, glucose distribution, BMI comparison, average glucose comparison and correlation heatmap.
6. Summarize findings and limitations in simple language.

> **Important:** This dataset is fully synthetic and the outcome label is simulated for learning. It is not a medical dataset of real patients, cannot diagnose diabetes, and must not guide treatment or health decisions.


```python
# 1. Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Display charts inside the notebook
%matplotlib inline
```

```python
# 2. Load the dataset
# Keep Medical_Dataset.csv in the same folder as this notebook.
df = pd.read_csv("Medical_Dataset.csv")
df.head()
```

```python
# 3. Basic dataset information
print("Shape:", df.shape)
print("\\nColumn information:")
df.info()
```

```python
# 4. Check missing values and duplicate rows
print("Missing values by column:")
print(df.isnull().sum())

print("\\nDuplicate rows:", df.duplicated().sum())
```

```python
# 5. Clean the data
clean_df = df.copy()

numeric_columns = [
    "BMI",
    "Cholesterol_mg_dL",
    "Physical_Activity_hours_week"
]

for column in numeric_columns:
    clean_df[column] = clean_df[column].fillna(clean_df[column].median())

clean_df["Family_History"] = clean_df["Family_History"].astype(int)
clean_df["Diabetes_Status"] = clean_df["Diabetes_Status"].astype(int)

print("Missing values after cleaning:")
print(clean_df.isnull().sum())
```

```python
# 6. Descriptive statistics
clean_df.describe().round(2)
```

```python
# 7. Count of each simulated outcome label
print(clean_df["Diabetes_Status"].value_counts())

clean_df["Diabetes_Status"].value_counts().sort_index().plot(
    kind="bar", title="Distribution of Simulated Outcome Labels"
)
plt.xlabel("Outcome label (0=lower-risk, 1=higher-risk)")
plt.ylabel("Number of records")
plt.xticks(rotation=0)
plt.show()
```

```python
# 8. Age distribution
plt.figure(figsize=(7, 4))
plt.hist(clean_df["Age"], bins=12, edgecolor="black")
plt.title("Patient Age Distribution")
plt.xlabel("Age (years)")
plt.ylabel("Number of records")
plt.show()
```

```python
# 9. Glucose distribution
plt.figure(figsize=(7, 4))
plt.hist(clean_df["Glucose_mg_dL"], bins=14, edgecolor="black")
plt.title("Glucose Level Distribution")
plt.xlabel("Glucose (mg/dL)")
plt.ylabel("Number of records")
plt.show()
```

```python
# 10. Compare BMI across outcome groups
clean_df.boxplot(column="BMI", by="Diabetes_Status")
plt.title("BMI by Simulated Outcome Label")
plt.suptitle("")
plt.xlabel("Outcome label (0=lower-risk, 1=higher-risk)")
plt.ylabel("BMI (kg/m²)")
plt.show()
```

```python
# 11. Compare average glucose across outcome groups
clean_df.groupby("Diabetes_Status")["Glucose_mg_dL"].mean().plot(
    kind="bar", title="Average Glucose by Simulated Outcome Label"
)
plt.xlabel("Outcome label")
plt.ylabel("Average glucose (mg/dL)")
plt.xticks(rotation=0)
plt.show()
```

```python
# 12. Correlation heatmap (using Matplotlib only)
correlation = clean_df.select_dtypes(include=np.number).corr()

plt.figure(figsize=(9, 6))
plt.imshow(correlation, aspect="auto", vmin=-1, vmax=1)
plt.colorbar(label="Correlation")
plt.xticks(range(len(correlation.columns)), correlation.columns,
           rotation=65, ha="right")
plt.yticks(range(len(correlation.columns)), correlation.columns)
plt.title("Correlation Heatmap of Numeric Variables")
plt.tight_layout()
plt.show()
```

```python
# 13. Compare group averages
group_means = clean_df.groupby("Diabetes_Status")[
    ["Age", "BMI", "Glucose_mg_dL", "Blood_Pressure_mmHg",
     "Cholesterol_mg_dL", "Physical_Activity_hours_week"]
].mean().round(2)

group_means
```

```python
# 14. Key summary
total_records = len(clean_df)
higher_risk_records = int(clean_df["Diabetes_Status"].sum())
lower_risk_records = total_records - higher_risk_records

print("Total records:", total_records)
print("Higher-risk label records:", higher_risk_records)
print("Lower-risk label records:", lower_risk_records)
print("Higher-risk label share:", round(higher_risk_records / total_records * 100, 2), "%")

print("\\nNote: These labels are simulated and are not clinical diagnoses.")
```
